```python
# Now, let's solve some questions to get more clarity of the dataset

# 1) How many unique YouTube categories exist?
# 2) Which 10 channels have the most subscribers?
# 3) Which categories have the highest average subscribers?
# 4) Which channels have more than 10 billion video views?
# 5) Which categories have the highest total video views?
# 6) Which country has the most YouTube channels?
# 7) Which countries have the highest average subscribers?
# 8) Which channels gained the most subscribers in the last 30 days?
# 9) Which categories have the highest estimated monthly earnings?
# 10) What is the average lowest monthly earnings estimate? 
```

```python
df = pd.read_csv('Cleaned_Global_youtube_statistics.csv')
```

```python
# 1) How many unique YouTube categories exist?
print(df['category'].nunique())
```

```python
# 2) Which 10 channels have the most subscribers?

top_10_subscribers = df.nlargest(10, 'subscribers')

print(top_10_subscribers[['Youtuber', 'subscribers']])
```

T-series youtube channel has the highest numbers of subribers 

```python
# 3) Which categories have the highest average subscribers?

avg_subribers = (df.groupby('category')['subscribers'].mean().sort_values())
print(avg_subribers)
```

Howto & Style category gets the most number of subscribers 

```python
# 4) Which channels have more than 10 billion video views?

highest_views = df[df['video views'] > 10_000_000_000]
print(highest_views[['Youtuber','video views']])
```

T-Series youtube channel have more than 10 Billion views

```python
# 5) Which categories have the highest total video views?

highest_total_views = (df.groupby('category')['video views']
                       .sum()
                       .sort_values(ascending=False)
                       )
print(highest_total_views)
```

Music category is getting highest number of views

```python
# 6) Which country has the most YouTube channels?

country = df['Country'].value_counts()

print(country.head(10))
```

United States has the highest number of youtube channels whereas Argentina and Spain have the lowest

```python
# 7) Which countries have the highest average subscribers?

country_avg_subscribers = (df.groupby('Country')['subscribers'].mean().sort_values(ascending=False))
print(country_avg_subscribers.head(10))
```

In average, Cuba have the highest number of subscribers 

```python
# 8) Which channels gained the most subscribers in the last 30 days?

top_recent_growth = df.nlargest(10,'subscribers_for_last_30_days')
print(top_recent_growth[['Youtuber','subscribers','subscribers_for_last_30_days']])
```

MrBeast has gained the most number of subscribers in the last 30 days

```python
# 9) Which categories have the highest estimated monthly earnings?

df['average_monthly_earnings'] = (
    df['lowest_monthly_earnings'] + df['highest_monthly_earnings']
)/2

category_earnings = (df.groupby('category')['average_monthly_earnings']
                     .mean().sort_values(ascending=False))

print(category_earnings)
```

In an average, Category 'Shows' earns highest in every month

```python
# 10) What is the average lowest monthly earnings estimate?

avg_lowest_earnings = df['lowest_monthly_earnings'].mean()
print(avg_lowest_earnings) 
```

```python
# Now, it's time for visualize the dataset 
# Let's start with matplotlib

# Bar chart showing the number the of channels in each category

category_counts = df['category'].value_counts()

plt.figure(figsize=(12,8))
category_counts.plot(kind='bar')

plt.title('Number of Youtube channels by Category')
plt.xlabel('Category')
plt.ylabel('Number of channels')
plt.xticks(rotation=45,ha='right')
plt.tight_layout()
plt.show()
```

The bar graph shows that Entertainment category has the highest number of Youtube channel 

```python
# Using Seaborn for horizontal bar chart 
# Showing top 10 Youtube channels by subscribers 

top_10_subscribers = df.nlargest(10,'subscribers')

plt.figure(figsize=(12,8))
sns.barplot(
    data = top_10_subscribers,
    x = 'subscribers',
    y='Youtuber'
)

plt.title('Top 10 Youtube Channels by Subscribers')
plt.xlabel('Subscribers')
plt.ylabel('Youtuber')
plt.tight_layout()
plt.show()
```

T-Series is the most subscribed channel

```python
import matplotlib.pyplot as plt
import pandas as pd

item_count = df['category'].value_counts()

# Select the top 10 categories
top_7 = item_count.head(7)

# Combine the remaining categories into Others
others = item_count.iloc[7:].sum()

pie_data = pd.concat([
    top_7,
    pd.Series({'Others': others})
])

plt.figure(figsize=(10, 8))

plt.pie(
    pie_data,
    labels=pie_data.index,
    autopct='%1.2f%%',
    startangle=90,
    pctdistance=0.75,
    labeldistance=1.1
)

plt.title('Distribution of YouTube Channels by Category')
plt.axis('equal')
plt.tight_layout()
plt.show()
```

Pie Chart shows Entertainment Category has the highest percentage of Youtube channels

```python
# Correlation Heatmap for the relationships between some numerical columns

selected_columns = [
    'subscribers',
    'video views',
    'uploads',
    'video_views_for_the_last_30_days',
    'lowest_monthly_earnings',
    'highest_monthly_earnings'
]

sns.heatmap(
    df[selected_columns].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Between Key YouTube Statistics')
plt.tight_layout()
plt.show()
```

Subscribers and video views have a strong relationship ,while the number of uploads has a weak relationship with most other statistics 

```python
# Distribution of lowest monthly earnings estimates with the help of histogram

average_lowest = df['lowest_monthly_earnings'].mean()

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='lowest_monthly_earnings',
    bins=30
)

plt.axvline(
    average_lowest,
    linestyle='--',
    label=f'Mean = {average_lowest:,.0f}'
)

plt.title('Distribution of Lowest Estimated Monthly Earnings')
plt.xlabel('Lowest Monthly Earnings Estimate')
plt.ylabel('Number of Channels')
plt.legend()
plt.tight_layout()
plt.show()
```

Most channels have relatively low estimated monthly earnings
