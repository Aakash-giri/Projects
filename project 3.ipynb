```python
# Moving to EDA (Exploratory Data Analysis)

import pandas as pd

df = pd.read_csv('Cleaned_netflix_titles.csv')

# Dataset dimensions
print(df.shape)
```

```python
# Column names and data types
print(df.info())
```

```python
# Numerical statistical summary
print(df.describe())
```

```python
df.columns
```

```python
# Checking multiple unique values of all columns
columns = [
    'title', 'director', 'country', 'date_added',
    'release_year', 'rating', 'duration',
    'listed_in', 'description'
]

for col in columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
# Correlation between numerical columns
print(df.select_dtypes(include='number').corr())
```

```python
# Some questions with visualizations for more clarity of the dataset
```

```python
# 1) Which countries have the highest number of movies (titles) on Netflix?

# Split multiple countries and count each country
country_count = (
    df['country']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
)

# Select the top 10 countries
top_10_countries = country_count.head(10)

print(top_10_countries)
```

```python
# Plot
top_10_countries.sort_values().plot(
    kind='barh',
    figsize=(10, 6)
)

plt.title('Top 10 Countries by Number of Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

United States has the most number of movies(titles)

```python
# What are the most common content ratings on Netflix?

# Count each content rating
rating_count = df['rating'].value_counts()

# Select the top 10 ratings
top_10_ratings = rating_count.head(10)

print(top_10_ratings)
```

```python
# Plot
top_10_ratings.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Top 10 Content Ratings on Netflix')
plt.xlabel('Content Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

The most common content rating is TV-MA

```python
# Which directors have the most titles on Netflix?

# Split and count individual directors
director_count = (
    df['director']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
)
# Select the top 10 directors
top_10_directors = director_count.head(10)

print(top_10_directors)
```

```python
# Plot
top_10_directors.sort_values().plot(
    kind='barh',
    figsize=(10, 6)
)

plt.title('Top 10 Directors by Number of Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Director')
plt.tight_layout()
plt.show()
```

Director, Jan Suter has directed most number of movies
