```python
# 1) Do restaurants with higher ratings also receive more ratings?

correlation = df[
    ['rate (out of 5)', 'num of ratings']
].corr()

print(correlation)
```

```python
# Creating a scatter plot 

sns.scatterplot(
    data=df,
    x='num of ratings',
    y='rate (out of 5)',
    alpha=0.7
   )


plt.title('Restaurant Ratings vs Number of Ratings')
plt.xlabel('Number of Ratings')
plt.ylabel('Rating (out of 5)')
plt.show()
```

The correlation between rating(out of 5) and Number of ratings is positive as both are increasing relatively together 

```python
# Seeing top 10 cuisine types 
df = pd.read_csv('Cleaned_zomato.csv')

cuisines_type = df['cuisines type'].value_counts()
top_10 = cuisines_type.head(10).sort_values(ascending=False)

top_10.plot(kind='barh', figsize=(10, 6))

plt.title('Top 10 Most Common Cuisine Types')
plt.xlabel('Number of Restaurants Listing the Cuisine')
plt.ylabel('Cuisine Type')
plt.show()
```

Most number of restaurants have listed North Indian, Chinese cuisine

```python
import matplotlib.pyplot as plt

# Find the top 10 most expensive restaurants
top_10_restaurants = df.nlargest(10, 'avg cost (two people)').copy()

# Sort costs from lowest to highest for a horizontal bar chart
top_10_restaurants = top_10_restaurants.sort_values(
    'avg cost (two people)', ascending=True
)

# Create bar chart
plt.figure(figsize=(12, 7))

plt.barh(
    top_10_restaurants['restaurant name'],
    top_10_restaurants['avg cost (two people)']
)

# Add cost labels to each bar
for i, cost in enumerate(top_10_restaurants['avg cost (two people)']):
    plt.text(cost, i, f' {cost:.0f}', va='center')

plt.title('Top 10 Most Expensive Restaurants')
plt.xlabel('Average Cost for Two People')
plt.ylabel('Restaurant Name')
plt.tight_layout()
plt.show()
```

 The chart shows which restaurants have the highest estimated dining costs for two people. Longer bars represent higher costs, making it easy to compare the restaurants.
